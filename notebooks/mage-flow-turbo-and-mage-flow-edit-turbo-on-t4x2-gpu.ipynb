## 01. Configuration
## 01. Cấu hình

**English**

Define public source/support paths, fixed device routing and reusable notebook
helpers. Source checkout, patch application, runtime restore and session-token
creation happen in later stages so this cell is safe in an empty
/kaggle/working directory.

**Tiếng Việt**

Khai báo đường dẫn public source/support, ánh xạ GPU cố định và các helper dùng
lại trong notebook. Việc checkout source, áp dụng patch, restore runtime và tạo
session token được thực hiện ở các stage sau, vì vậy cell này an toàn ngay cả
khi /kaggle/working ban đầu trống.



```python
from pathlib import Path
import base64, hashlib, io, json, os, subprocess, sys, threading, time
import urllib.request

PROJECT_ROOT = Path("/kaggle/working/mage-flow-turbo-and-mage-flow-edit-turbo-on-t4x2-gpu").resolve()
RUNTIME_ROOT = Path("/kaggle/working/mage-flow-t4x2-runtime").resolve()
SUPPORT_ROOT = Path("/kaggle/input/datasets/dangkhoa2016/mage-flow-t4x2-runtime-support").resolve()
RUNTIME_ARCHIVE = SUPPORT_ROOT / "runtime" / "mage-flow-t4x2-runtime-py312-torch213-cu126.tar.zst"
RUNTIME_ARCHIVE_SHA256 = "4f59c76fffa79c86d19ca3ae872dd80163fda38f625f3a125da160b363699b32"
OPTIMIZATION_PATCH = SUPPORT_ROOT / "qualification" / "patches" / "pinned-memory-offload-optimization.patch"
OPTIMIZATION_PATCH_SHA256 = "89eb48e13914c9d3aa422360ac55355b821ad9beb41c2ffa491571eade682506"
PUBLIC_REPO_URL = "https://github.com/dangkhoa2016/Mage-Flow-Turbo-And-Mage-Flow-Edit-Turbo-On-T4x2-GPU.git"
PUBLIC_BRANCH = "main"
RUNTIME_PYTHON = RUNTIME_ROOT / ".venv" / "bin" / "python"
MAGE_SOURCE = RUNTIME_ROOT / "vendor" / "Mage"
EXPECTED_MAGE_COMMIT = "76bec2bb3818863f470de7e867c2dc7f1d0bfd83"
T2I_MODEL = Path("/kaggle/input/models/dangkhoa2016/mage-flow-community-mage-flow-turbo/pytorch/default/1")
EDIT_MODEL = Path("/kaggle/input/models/dangkhoa2016/mage-flow-community-mage-flow-edit-turbo/pytorch/default/1")
GGUF_ROOT = Path("/kaggle/input/models/dangkhoa2016/qwen-qwen3-vl-4b-instruct-gguf/gguf/q4-k-m/1")
EDIT_SOURCE = EDIT_MODEL / "assets" / "dog.jpg"
REST_HOST = "127.0.0.1"
REST_PORT = 8090
REST_URL = f"http://{REST_HOST}:{REST_PORT}"
T2I_DEVICE = "cuda:0"
EDIT_DEVICE = "cuda:1"
OUTPUT_DIR = PROJECT_ROOT / "outputs"
ACCEPT_DIR = PROJECT_ROOT / ".runtime" / "acceptance"
BAR = "━" * 46
TOKEN = None
PUBLIC_SOURCE_COMMIT = None

def stage(label: str) -> None:
    print(BAR); print(f"[STAGE] {label}"); print(BAR)

def heartbeat(msg: str) -> None:
    print(f"[HEARTBEAT] {msg}", flush=True)

def sha256_file(path: Path, chunk_size: int = 8 * 1024 * 1024) -> str:
    h = hashlib.sha256()
    with path.open("rb") as fh:
        while True:
            chunk = fh.read(chunk_size)
            if not chunk:
                break
            h.update(chunk)
    return h.hexdigest()

def run_cmd(cmd, label: str, heartbeat_seconds: int = 45, cwd=None) -> int:
    stage(label)
    proc = subprocess.Popen(cmd, cwd=cwd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                            text=True, bufsize=1)
    stop_hb = threading.Event()
    started = time.monotonic()
    def _hb() -> None:
        while not stop_hb.wait(heartbeat_seconds):
            heartbeat(f"{label} still running elapsed={int(time.monotonic() - started)}s")
    hb = threading.Thread(target=_hb, daemon=True)
    hb.start()
    try:
        try:
            for line in proc.stdout:
                print(line, end="", flush=True)
        except Exception as exc:
            print(f"[WARN] output stream error: {type(exc).__name__}: {exc}", flush=True)
        proc.wait()
    finally:
        stop_hb.set()
        if proc.stdout is not None:
            proc.stdout.close()
    if proc.returncode != 0:
        raise RuntimeError(f"command failed rc={proc.returncode}: {' '.join(map(str, cmd))}")
    print(f"[PASS] {label} rc=0")
    return 0

def resolve_token():
    token_file = PROJECT_ROOT / ".runtime" / "api_token"
    import sys as _sys
    if str(PROJECT_ROOT) not in _sys.path:
        _sys.path.insert(0, str(PROJECT_ROOT))
    from scripts.token_store import ensure_token as _ensure_token
    token = _ensure_token(token_file)
    os.environ["MAGE_FLOW_API_TOKEN"] = token
    return token

def api_headers() -> dict:
    if not TOKEN:
        raise RuntimeError("session token not initialized")
    return {"Authorization": f"Bearer {TOKEN}"}

def sha256_hex(raw: bytes) -> str:
    return hashlib.sha256(raw).hexdigest()

def decode_data_url(data_url: str) -> bytes:
    _head, _, b64 = data_url.partition(",")
    return base64.b64decode(b64)

print(f"[INFO] PROJECT_ROOT={PROJECT_ROOT}")
print(f"[INFO] SUPPORT_ROOT={SUPPORT_ROOT}")
print(f"[INFO] REST={REST_URL}")
print(f"[INFO] T2I_DEVICE={T2I_DEVICE} EDIT_DEVICE={EDIT_DEVICE}")
if not (T2I_DEVICE == "cuda:0" and EDIT_DEVICE == "cuda:1"):
    raise RuntimeError("device configuration mismatch: cuda:0/cuda:1 required")
print("[PASS] NOTEBOOK_CONFIGURATION")

```

Kiểm tra môi trường clean session, clone public repository từ main, ghi lại

```python
import platform, shutil, sys
stage("Clean-session bootstrap + Kaggle environment preflight")
print(f"[INFO] python={sys.version.split()[0]}")
print(f"[INFO] platform={platform.platform()}")
for p in ("/kaggle/input", "/kaggle/working"):
    print(f"[INFO] exists {p}={Path(p).exists()}")
if not Path("/kaggle/working").exists():
    raise RuntimeError("/kaggle/working is unavailable")
usage = shutil.disk_usage("/kaggle/working")
print(f"[INFO] working_disk_gb used={usage.used/2**30:.2f} free={usage.free/2**30:.2f}")
if not SUPPORT_ROOT.is_dir():
    raise RuntimeError(f"support Dataset not attached: {SUPPORT_ROOT}")
if not OPTIMIZATION_PATCH.is_file():
    raise RuntimeError(f"optimization patch missing: {OPTIMIZATION_PATCH}")
patch_sha = sha256_file(OPTIMIZATION_PATCH)
print(f"[INFO] optimization_patch_sha256={patch_sha}")
if patch_sha != OPTIMIZATION_PATCH_SHA256:
    raise RuntimeError(f"optimization patch SHA mismatch: {patch_sha}")

if PROJECT_ROOT.exists():
    if not (PROJECT_ROOT / ".git").is_dir():
        raise RuntimeError(f"PROJECT_ROOT exists but is not a Git checkout: {PROJECT_ROOT}")
    print("[INFO] refreshing existing public source checkout to latest main")
    run_cmd(
        ["git", "-C", str(PROJECT_ROOT), "fetch", "--depth", "1", "origin", PUBLIC_BRANCH],
        "Fetch latest public source",
        heartbeat_seconds=15,
    )
    run_cmd(
        ["git", "-C", str(PROJECT_ROOT), "reset", "--hard", "FETCH_HEAD"],
        "Reset public source to fetched main",
        heartbeat_seconds=15,
    )
else:
    run_cmd(
        ["git", "clone", "--depth", "1", "--branch", PUBLIC_BRANCH, PUBLIC_REPO_URL, str(PROJECT_ROOT)],
        "Clone public source",
        heartbeat_seconds=15,
    )

PUBLIC_SOURCE_COMMIT = subprocess.run(
    ["git", "-C", str(PROJECT_ROOT), "rev-parse", "HEAD"],
    check=True, capture_output=True, text=True,
).stdout.strip()
print(f"[INFO] public_source_commit={PUBLIC_SOURCE_COMMIT}")

diff_names = subprocess.run(
    ["git", "-C", str(PROJECT_ROOT), "diff", "--name-only"],
    check=True, capture_output=True, text=True,
).stdout.splitlines()
if not diff_names:
    subprocess.run(
        ["git", "-C", str(PROJECT_ROOT), "apply", "--check", str(OPTIMIZATION_PATCH)],
        check=True,
    )
    subprocess.run(
        ["git", "-C", str(PROJECT_ROOT), "apply", str(OPTIMIZATION_PATCH)],
        check=True,
    )
    diff_names = subprocess.run(
        ["git", "-C", str(PROJECT_ROOT), "diff", "--name-only"],
        check=True, capture_output=True, text=True,
    ).stdout.splitlines()
if diff_names != ["server/workers/conditioning_offload.py"]:
    raise RuntimeError(f"unexpected public source diff after bootstrap: {diff_names}")
print("[PASS] PUBLIC_SOURCE_PLUS_PINNED_OPTIMIZATION_PATCH")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
ACCEPT_DIR.mkdir(parents=True, exist_ok=True)
TOKEN = resolve_token()
print("[INFO] MAGE_FLOW_API_TOKEN=<temporary session token; not printed>")
print("[PASS] CLEAN_SESSION_BOOTSTRAP")

```

## 05. Runtime restore / validation / reuse
## 05. Restore / xác minh / tái sử dụng runtime

**English**

On a clean session, verify the sealed runtime archive SHA-256 and stream-extract
it from the attached support Dataset into /kaggle/working. On a rerun, reuse an
existing runtime only after its Python executable, Mage source and pinned commit
all validate. A stale or partial runtime is removed and restored automatically
before any model load.

**Tiếng Việt**

Trong clean session, xác minh SHA-256 của sealed runtime archive rồi
stream-extract runtime từ support Dataset vào /kaggle/working. Khi chạy lại,
chỉ tái sử dụng runtime hiện có sau khi Python executable, Mage source và pinned
commit đều hợp lệ. Runtime stale/partial sẽ tự được xóa và restore lại trước khi
load model.



```python
stage("Runtime restore / validation / reuse")
import shutil

def validate_existing_runtime():
    if not RUNTIME_ROOT.is_dir():
        return False, "runtime root absent"
    if not RUNTIME_PYTHON.is_file():
        return False, f"runtime Python missing: {RUNTIME_PYTHON}"
    if not os.access(RUNTIME_PYTHON, os.X_OK):
        return False, f"runtime Python is not executable: {RUNTIME_PYTHON}"
    if not MAGE_SOURCE.is_dir():
        return False, f"Mage source missing: {MAGE_SOURCE}"
    try:
        mage_commit = subprocess.run(
            ["git", "-C", str(MAGE_SOURCE), "rev-parse", "HEAD"],
            capture_output=True, text=True, check=True,
        ).stdout.strip()
    except subprocess.CalledProcessError as exc:
        return False, f"Mage commit check failed rc={exc.returncode}"
    if mage_commit != EXPECTED_MAGE_COMMIT:
        return False, f"Mage commit mismatch: {mage_commit}"
    try:
        identity = subprocess.run(
            [str(RUNTIME_PYTHON), "-c", "import sys; print(sys.version.split()[0])"],
            capture_output=True, text=True, timeout=30,
        )
    except (OSError, subprocess.SubprocessError) as exc:
        return False, f"runtime Python launch failed: {type(exc).__name__}: {exc}"
    if identity.returncode != 0:
        return False, f"runtime Python identity rc={identity.returncode}"
    return True, identity.stdout.strip()

reuse_ok, reuse_detail = validate_existing_runtime()
if RUNTIME_ROOT.exists() and not reuse_ok:
    print(f"[WARN] stale/partial runtime detected; restoring clean runtime: {reuse_detail}")
    shutil.rmtree(RUNTIME_ROOT)

if not reuse_ok:
    if not RUNTIME_ARCHIVE.is_file():
        raise RuntimeError(f"runtime archive missing: {RUNTIME_ARCHIVE}")
    actual_archive_sha = sha256_file(RUNTIME_ARCHIVE)
    print(f"[INFO] runtime_archive_sha256={actual_archive_sha}")
    if actual_archive_sha != RUNTIME_ARCHIVE_SHA256:
        raise RuntimeError(f"runtime archive SHA mismatch: {actual_archive_sha}")
    import tarfile, zstandard as zstd
    stage("Restore validated runtime")
    started = time.monotonic()
    last_report = started
    with RUNTIME_ARCHIVE.open("rb") as src:
        with zstd.ZstdDecompressor().stream_reader(src) as reader:
            with tarfile.open(fileobj=reader, mode="r|") as tf:
                for member in tf:
                    tf.extract(member, path="/kaggle/working", filter="fully_trusted")
                    now = time.monotonic()
                    if now - last_report >= 45:
                        heartbeat(f"runtime restore still running elapsed={int(now - started)}s")
                        last_report = now
    print(f"[PASS] VALIDATED_RUNTIME_RESTORED elapsed={time.monotonic() - started:.1f}s")
else:
    print(f"[INFO] validated existing runtime detected; extraction skipped python={reuse_detail}")

if not RUNTIME_PYTHON.is_file():
    raise RuntimeError(f"validated runtime Python missing: {RUNTIME_PYTHON}")
if not os.access(RUNTIME_PYTHON, os.X_OK):
    raise RuntimeError(f"validated runtime Python is not executable: {RUNTIME_PYTHON}")
if not MAGE_SOURCE.is_dir():
    raise RuntimeError(f"Mage source missing: {MAGE_SOURCE}")
actual = subprocess.run(
    ["git", "-C", str(MAGE_SOURCE), "rev-parse", "HEAD"],
    capture_output=True, text=True, check=True,
).stdout.strip()
print(f"[INFO] runtime_python={RUNTIME_PYTHON}")
print(f"[INFO] mage_commit={actual}")
if actual != EXPECTED_MAGE_COMMIT:
    raise RuntimeError(f"commit mismatch: expected {EXPECTED_MAGE_COMMIT} found {actual}")
identity = subprocess.run(
    [str(RUNTIME_PYTHON), "-c", "import sys; print(sys.version.split()[0])"],
    capture_output=True, text=True,
)
print(f"[INFO] runtime python version: {identity.stdout.strip()}")
if identity.returncode != 0:
    raise RuntimeError("runtime Python identity check failed")
print("[PASS] RUNTIME_RESTORE_VALIDATION_REUSE")
```

## 06. Split-device configuration

## 09. Wait for /ready
## 09. Chờ /ready

**English**

Poll the authenticated `GET /ready` endpoint with heartbeat output. `/ready`
becomes `true` **only** when the coordinator, the T2I worker and the Edit worker
are all ready on their required devices. A bounded wait prevents endless loops.

**Tiếng Việt**

Poll endpoint `GET /ready` (có xác thực) kèm heartbeat. `/ready` chỉ trở thành
`true` khi **tất cả** coordinator, T2I worker và Edit worker sẵn sàng trên đúng
thiết bị yêu cầu. Giới hạn thời gian chờ tránh vòng lặp vô tận.


```python
stage("Wait for /ready")
import time as _time
wait_started = _time.monotonic()
deadline = wait_started + 300
last_report = 0.0
last_payload = {}
while _time.monotonic() < deadline:
    r = requests.get(f"{REST_URL}/ready", headers=api_headers(), timeout=10)
    last_payload = r.json()
    if r.status_code == 200 and last_payload.get("ready") is True:
        break
    now = _time.monotonic()
    if now - last_report >= 15:
        heartbeat(f"waiting /ready elapsed={int(now - wait_started)}s status={last_payload.get('status')}")
        last_report = now
    _time.sleep(3)
else:
    raise RuntimeError(f"/ready did not become true: {last_payload}")
print(f"[INFO] /ready -> {json.dumps(last_payload, indent=2)}")
if not (last_payload.get("t2i_ready") is True):
    raise RuntimeError('last_payload.get("t2i_ready") is True')
if not (last_payload.get("edit_ready") is True):
    raise RuntimeError('last_payload.get("edit_ready") is True')
print("[PASS] REST_API_READY")
```

## 16. Optional Quick Tunnel
## 16. Quick Tunnel tùy chọn

**English**

Optionally expose the localhost coordinator through a temporary Cloudflare Quick
Tunnel. Only port `8090` is tunnelled; the internal worker ports `8101`/`8102`
are never exposed. The public boundary stays protected by the per-session Bearer
token. This is a demo convenience, not a permanent or SLA-backed endpoint. It
runs only when `ENABLE_QUICK_TUNNEL=1`.

**Tiếng Việt**

Tùy chọn expose coordinator localhost qua Cloudflare Quick Tunnel tạm thời. Chỉ
port `8090` được tunnel; hai port worker nội bộ `8101`/`8102` không bao giờ
được expose. Boundary public vẫn được bảo vệ bằng Bearer token của session. Đây
chỉ là tiện ích demo, không phải endpoint public lâu dài hay có SLA. Chỉ chạy
khi `ENABLE_QUICK_TUNNEL=1`.


```python
stage("Optional Quick Tunnel")
TUNNEL_URL = None
if os.environ.get("ENABLE_QUICK_TUNNEL") == "1":
    import shutil as _shutil
    if not (_shutil.which("cloudflared")):
        raise RuntimeError("cloudflared not installed; cannot start tunnel")
    tunnel_log = PROJECT_ROOT / ".runtime" / "tunnel.log"
    proc = subprocess.Popen(
        ["cloudflared", "tunnel", "--url", REST_URL, "--no-autoupdate"],
        stdout=open(tunnel_log, "a"), stderr=subprocess.STDOUT)
    print(f"[INFO] cloudflared started pid={proc.pid} log={tunnel_log}")
    deadline = _time.monotonic() + 120
    while _time.monotonic() < deadline:
        text = tunnel_log.read_text() if tunnel_log.is_file() else ""
        import re as _re
        m = _re.search(r"https://[a-z0-9-]+\.trycloudflare\.com", text)
        if m:
            TUNNEL_URL = m.group(0)
            break
        _time.sleep(3)
    if not (TUNNEL_URL):
        raise RuntimeError("tunnel URL not detected within 120s; see .runtime/tunnel.log")
    print(f"[INFO] TUNNEL_URL={TUNNEL_URL}")
    print("[INFO] tunnel exposes coordinator only; worker ports remain localhost-only")
    print("[PASS] QUICK_TUNNEL_STARTED")
else:
    print("[INFO] ENABLE_QUICK_TUNNEL!=1 -> optional tunnel skipped (resident workers untouched)")
    print("[PASS] QUICK_TUNNEL_SKIPPED")
```
